```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

# Dataset: CO2 Emissions by Country 2000-2022
# Source: Our World in Data (https://ourworldindata.org/co2-emissions)
df = pd.read_csv('../data/co2_emissions.csv')
print(f"Loaded: {len(df)} rows | Countries: {df['Country'].nunique()} | Years: {df['Year'].min()}-{df['Year'].max()}")
print(df.head())
```

```python
# Explore before building

print("Countries:", df['Country'].unique())
print("\nCO2 range:", df['CO2_Mt'].min(), "to", df['CO2_Mt'].max(), "Mt")
print("\nRegional averages (2022):")
print(df[df['Year']==2022].groupby('Region')['CO2_Mt'].mean().sort_values(ascending=False).round(1))
```

## Task — Multi-Series Line Chart with Highlight

**What to build:** A line chart showing CO2 emissions over time for **all Asian countries** in the dataset, with one country highlighted.

**Requirements:**
- All countries shown (for context), but only **one highlighted in colour** — your choice which
- All other lines in grey (#EEEEEE), thinner
- Highlighted country **labelled directly** at the end of its line (not in a legend)
- Insight title that names the highlighted country and its story

```python
# Task — Multi-series line with highlight
# Lecture 3: CO2 emissions in Asian countries
# --------------------------------------------

# Select Asian countries
asia = df[df['Region'].str.contains('Asia', case=False, na=False)].copy()

# Choose the highlighted country
highlight_country = 'China'

if asia.empty:
    print("No Asian countries found. Check the Region column values.")
elif highlight_country not in asia['Country'].unique():
    print("China was not found. Check the Country column values.")
else:
    fig = go.Figure()

    # Add all other countries in light grey
    for country in asia['Country'].unique():
        if country == highlight_country:
            continue

        country_data = asia[asia['Country'] == country].sort_values('Year')

        fig.add_trace(go.Scatter(
            x=country_data['Year'],
            y=country_data['CO2_Mt'],
            mode='lines',
            line=dict(color='#EEEEEE', width=1),
            hovertemplate=(
                country + '<br>Year: %{x}<br>CO₂: %{y:.1f} Mt'
                '<extra></extra>'
            ),
            showlegend=False
        ))

    # Add China in colour
    china = asia[asia['Country'] == highlight_country].sort_values('Year')

    fig.add_trace(go.Scatter(
        x=china['Year'],
        y=china['CO2_Mt'],
        mode='lines',
        line=dict(color='#D94841', width=3),
        name=highlight_country,
        showlegend=False,
        hovertemplate=(
            'China<br>Year: %{x}<br>CO₂: %{y:.1f} Mt'
            '<extra></extra>'
        )
    ))

    # Label China at the end of its line
    last_year = china.iloc[-1]

    fig.add_annotation(
        x=last_year['Year'],
        y=last_year['CO2_Mt'],
        text='China',
        showarrow=False,
        xshift=30,
        font=dict(size=13, color='#D94841')
    )

    fig.update_layout(
        title='China’s CO₂ Emissions Rise Above the Regional Background',
        xaxis_title='Year',
        yaxis_title='CO₂ emissions (Mt)',
        plot_bgcolor='white',
        paper_bgcolor='white',
        showlegend=False,
        height=600,
        margin=dict(l=60, r=100, t=90, b=60),
        xaxis=dict(showline=False, showgrid=False, zeroline=False),
        yaxis=dict(showline=False, showgrid=False, zeroline=False)
    )

    fig.show()
```
